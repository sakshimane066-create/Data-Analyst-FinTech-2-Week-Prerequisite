```python
import glob
keys = [
    "01_fund_master", "02_nav_history", "03_aum_by_fund_house",
    "04_monthly_sip_inflows", "05_category_inflows", "06_industry_folio_count",
    "07_scheme_performance", "08_investor_transactions",
    "09_portfolio_holdings", "10_benchmark_indices",
]

datasets = {}
for k in keys:
    matches = sorted(glob.glob(f"*{k}*.csv"))
    if not matches:
        print(f"[!] No file found for {k} - upload it to the same folder as this notebook")
        continue
    datasets[k] = pd.read_csv(matches[0])
print(f"Loaded {len(datasets)} of {len(keys)} datasets\n")

for name, df in datasets.items():
    print("=" * 70)
    print(name, df.shape)
    print("=" * 70)

    nulls = df.isna().sum()
    nulls = nulls[nulls > 0]
    if nulls.empty:
        print("1. Missing values : none")
    else:
        print("1. Missing values :")
        for col, n in nulls.items():
            print(f"     {col}: {n} ({n / len(df) * 100:.1f}%)")

    print("2. Duplicate rows :", int(df.duplicated().sum()))

    text_dates = [
        c for c in df.columns
        if any(k in c.lower() for k in ("date", "month"))
        and not pd.api.types.is_datetime64_any_dtype(df[c])
    ]
    print("3. Dates stored as text :", text_dates if text_dates else "none")

    num = df.select_dtypes(include="number")
    neg = (num < 0).sum()
    neg = neg[neg > 0]
    print("4. Negative values :", "none" if neg.empty else neg.to_dict())

    outliers = {}
    for c in num.columns:
        if "code" in c.lower() or c.lower().endswith("_id"):
            continue
        q1, q3 = num[c].quantile([0.25, 0.75])
        iqr = q3 - q1
        if iqr == 0:
            continue
        n = int(((num[c] < q1 - 3 * iqr) | (num[c] > q3 + 3 * iqr)).sum())
        if n:
            outliers[c] = n
    print("5. Outliers (check manually) :", outliers if outliers else "none")
    print()
```

## Anomalies noted

**Missing values**
- 04_monthly_sip_inflows: yoy_growth_pct has 12 NaN (25%). Expected, the first 12 months (2022) have no previous year to compare against.
- No missing values in the other 9 datasets.

**Duplicate rows**
- None in any of the 10 datasets.

**Dates stored as text (object), need pd.to_datetime**
- 01_fund_master: launch_date
- 02_nav_history, 03_aum_by_fund_house, 10_benchmark_indices: date
- 04_monthly_sip_inflows, 05_category_inflows, 06_industry_folio_count: month
- 08_investor_transactions: transaction_date
- 09_portfolio_holdings: portfolio_date

**Negative values**
- 07_scheme_performance: max_drawdown_pct is negative in all 40 rows. Expected, a drawdown is a loss, so this is not an error.
- No negatives anywhere else (no negative NAV or amounts).

**Outliers (reviewed)**
- 02_nav_history: nav flagged on 1,477 rows, but no daily move above 10% for any scheme. Not an error, it only reflects different NAV levels across schemes.
- 04_monthly_sip_inflows: new_sip_accounts_lakh = 46.0 in 2025-04, about 4x the usual ~10 lakh. Suspicious spike, needs verification.
- 05_category_inflows: all 12 flagged values are the Liquid category (32k-42k crore a month). Expected, as liquid funds have much larger inflows than other categories.
- 07_scheme_performance: beta (max 1.04), sharpe_ratio and sortino_ratio flagged only because debt and equity schemes sit on different scales. Values look realistic.
- 09_portfolio_holdings covers only 34 of the 40 schemes in fund_master. The 6 missing are all Debt schemes (2 Gilt, 1 Short Duration, 3 Liquid), which hold no stocks. Expected, not an error.
- No outliers in the other datasets.

```python
nav = datasets["02_nav_history"].copy()
nav["date"] = pd.to_datetime(nav["date"])
nav = nav.sort_values(["amfi_code", "date"])
nav["daily_return_pct"] = nav.groupby("amfi_code")["nav"].pct_change() * 100
print("NAV: daily moves bigger than 10% (per scheme):")
print(nav[nav["daily_return_pct"].abs() > 10])

print("\nSIP new accounts, top 3:")
print(datasets["04_monthly_sip_inflows"].nlargest(3, "new_sip_accounts_lakh"))

print("\nCategory inflows, top 12:")
print(datasets["05_category_inflows"].nlargest(12, "net_inflow_crore"))

print("\nScheme performance, top beta:")
print(datasets["07_scheme_performance"].nlargest(6, "beta")[["scheme_name", "beta", "sharpe_ratio", "sortino_ratio"]])

print("\nPortfolio holdings, top weights:")
print(datasets["09_portfolio_holdings"].nlargest(3, "weight_pct"))
```
